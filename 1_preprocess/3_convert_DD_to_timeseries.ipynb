```python
import pandas as pd

input_file = ...
output_file = ...

df = pd.read_csv(input_file)

df_long = df.melt(
    id_vars=["pk_anio", "pk_mes"],
    var_name="day",
    value_name="ngasto"
)

df_long["day"] = df_long["day"].str.extract(r"(\d+)$").astype(int)

df_long["datetime"] = pd.to_datetime(
    dict(
        year=df_long["pk_anio"],
        month=df_long["pk_mes"],
        day=df_long["day"]
    ),
    errors="coerce"
)

df_long = (
    df_long
    .dropna(subset=["datetime"])
    .set_index("datetime")[["ngasto"]]
    .sort_index()
)

df_long.to_csv(output_file, index=True)
```
