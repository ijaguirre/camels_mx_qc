```python
import pandas as pd
from pathlib import Path

base_folder = Path(
    "/Users/iaguirre/Documents/camels_mx_wip/bandas_dataset/Bases_Datos_Bandas_CSV"
)

successful_files = []
failed_files = []

for station_folder in sorted(base_folder.iterdir()):
    if not station_folder.is_dir():
        continue

    for csv_path in station_folder.glob("HD*.csv"):
        print(f"Processing: {csv_path}")

        try:
            df = pd.read_csv(csv_path)

            #df["datetime"] = (
            #    pd.to_datetime(df["pk_fecha"], format="%m/%d/%y %H:%M:%S")
            #    + pd.to_timedelta(df["pk_hora"], unit="h")
            #)
            dates = pd.to_datetime(df["pk_fecha"], format="%m/%d/%y %H:%M:%S")

            dates = dates.where(
                dates.dt.year <= 2026,
                dates - pd.DateOffset(years=100)
            )

            df["datetime"] = dates + pd.to_timedelta(df["pk_hora"], unit="h")
            
            df = df.set_index("datetime")[["ngasto"]].sort_index()

            parquet_path = csv_path.with_suffix(".parquet")
            df.to_parquet(parquet_path)

            successful_files.append(str(csv_path))

        except Exception as e:
            failed_files.append({
                "file": str(csv_path),
                "error": str(e)
            })

print(f"\nSuccessful: {len(successful_files)}")
print(f"Failed: {len(failed_files)}")
```
