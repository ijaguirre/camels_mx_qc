# This code process the Bandas dataset to CSV

```python
import subprocess
from pathlib import Path
```

```python
input_folder = Path(
    "/Users/iaguirre/Documents/camels_mx_wip/bandas_dataset/Bases_Datos_Bandas"
)

output_folder = input_folder.parent / "Bases_Datos_Bandas_CSV"
output_folder.mkdir(parents=True, exist_ok=True)
```

```python
mdb_tables = "/opt/homebrew/bin/mdb-tables"
mdb_export = "/opt/homebrew/bin/mdb-export"

for database_path in sorted(input_folder.glob("*.mdb")):

    station_folder = output_folder / database_path.stem
    station_folder.mkdir(parents=True, exist_ok=True)

    tables = subprocess.check_output(
        [mdb_tables, "-1", str(database_path)],
        text=True
    ).splitlines()

    for table in tables:
        csv_path = station_folder / f"{table}.csv"

        with open(csv_path, "w", encoding="utf-8") as f:
            subprocess.run(
                [mdb_export, str(database_path), table],
                stdout=f,
                check=True
            )

        print(f"Saved: {csv_path}")

print("All databases converted.")
```
